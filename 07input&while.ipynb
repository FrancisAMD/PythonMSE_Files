### 进行检查的方法

```python
index = 0
input_id = " "
sample_id = []
while input_id != 'quit':
    input_id = input("请输入编号，输入quit以结束： ")
    if input_id != 'quit':
        sample_id.append(input_id)
        print(f'已登记ID{index+1}：{sample_id[index]}')
        index += 1
    else:
        print('已结束')
```

```python
I,J = 0,0
while I < 4:
    J = 0
    while J < 3:
        J += 1
        if J > 2:
            break
        print(I,J)
    I += 1
```

### 双层搜索结束

```python
groups = [[7.8, 0.0], [2.7]]
found = False
for group in groups:
    for mass in group:
        if mass <= 0:
            found = True
            break
    if found:
        break
print(found)
```

### 通过单独的 else 也能实现结束？

```python
masses_g = [7.8, 2.7]
for mass in masses_g:
    if mass >= 0:
        continue
    print("异常，发现非正的值")
else:
    print("正常，未发现非正的值")
```
