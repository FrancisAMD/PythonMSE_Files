用默认的空字符串省略附加的标签，进行判断

```python
def sample_label(sid, state, batch=""):  # 默认为空，也就是None
    label = f"{sid}: {state}"
    if batch: # 如果之前的时候输入过这是哪一批，就是batch不是空（None），这样可以给出一个新的标签，如果batch的输入参数是空的，那就是上面的那种标签，实现组别的标号
        label = f"{batch}/{label}"
    return label

print(sample_label("S01", "annealed"))
print(sample_label("S02", "quenched", "B02"))
```

一次返回多个值，实际返回一个元组对象

```python
def mean_and_count(values):
    return (
        sum(values) / len(values),
        len(values), # 这里还有一个逗号，对于tuple元组是可以的，如果是单元素的元组，这个逗号是必须的，没有逗号的话是一个int或者string或float
    )
result = mean_and_count([200, 210, 220])
average, count = result
print(result)
print(type(result).__name__)
print(average, count)
```

在函数中修改列表：任务转移

```python
def move_samples(pending, completed):
    while pending:
        completed.append(pending.pop())

pending = ["S01", "S02", "S03"]
completed = []
move_samples(pending, completed)
print(pending)
print(completed)
```

```python
def move_samples(pending, completed):
    while pending:
        completed.append(pending.pop())

original = ["S01", "S02"]
completed = []
move_samples(original[:], completed) # 只是类似于“浅拷贝”地，只读地输入original的数值，保留原列表并且进行操作
print(original)
print(completed)
```

用None默认值明确何时创建新列表

没有传入容器时新建，显式传入时按约定修改该容器

结果分别为['S01']与['S02']，对象不同

用is None判断，if not records会把调用者传入的空列表也换掉

文档明确是否修改传入容器，始终构造新结果可避免这种副作用

```python
def add_sample(sid, records = None):
    if records is None:
        records = []
    records.append(sid)
    return records

first = add_sample("S01")
second = add_sample("S02")
print(first, second)
print(first is second)
```

## 传递任意数量的实参


```python
def show_readings(*readings): # 允许多个参数 * 读进来的时候是一个tuple
    print(type(readings).__name__)
    print(readings)

show_readings(210.0)
show_readings(208.0, 210.0, 212.0)
```

```python
def show_readings(*readings):
    for index, value in enumerate(
        readings, start=1
    ):
        print(
            f"第{index}次：{value:.1f} HV" # HV 是维氏硬度
        )
show_readings(208.0, 210.0, 212.0)
```

```python
def reading_summary(sid, *readings):
    if not readings:
        return {
            "sample_id": sid,
            "count": 0,
        }
    return {
        "sample_id": sid,
        "mean_hv": sum(readings)
        / len(readings),
    }
print(reading_summary("S01", 208, 210, 212)) # 默认智能地认为是第一个是sid，后面都是tuple
```

在*readings之后声明unit；省略时使用默认值，覆盖时必须写出unit名称；覆盖也是非常方便的

```python
def show_readings(sid, *readings, unit="HV"):
    print(sid, readings, unit)

show_readings("RA", 310, 300, unit="HR")
```

**给到极高的自由度，但是要慎用

metadata是字典，收集没有匹配到其他形参的关键字实参，kwargs只是常见名称

本例仍应约定保留字段，避免调用者同时传入会被覆盖的sample_id

任意关键字提高灵活性，也减少拼写检查能力；核心材料字段适合用显式形参


```python
def build_sample(sid, **metadata):
    metadata["sample_id"] = sid
    return metadata

sample = build_sample(
    "S01", state="annealed", temperature_c=550
)
print(sample)
```

## 名称与作用域

```python
层次	英文	例子	建议
局部L	Local	形参、函数内中间值	优先用局部数据表达当前任务
外层E	Enclosing	嵌套函数中的外层名称	闭包作为提高内容，先能识别
全局G	Global	模块中的单位常量	减少可变全局状态依赖
内置B	Built-in	sum、list、input	不使用同名变量覆盖它们
```

两位同学都把变量叫 mass_g，不代表他们正在修改同一份局部记录

```python
mass_g = 7.8
def show_mass():
    mass_g = 2.7
    print("函数内", mass_g)
show_mass()
print("函数外", mass_g)
```

global 改的是模块中的名称绑定

```python
count = 0
def register():
    global count
    count += 1
register()
register()
print(count)
```

nonlocal：更新外层函数的状态


```python
def make_counter():
    count = 0
    def next_id():
        nonlocal count
        count += 1
        return count
    return next_id
next_id = make_counter()
print(next_id(), next_id())
```

##  函数对象与进阶用法

```python
def hardness_key(record):
    return record["hardness_hv"]

rows = [{"id": "S01", "hardness_hv": 220},
        {"id": "S02", "hardness_hv": 200}]
ordered = sorted(rows, key=hardness_key)
print([r["id"] for r in ordered])
print([r["id"] for r in rows])
```

lambda 形参: 表达式 会产生函数对象；表达式结果就是返回值

与 def 一样，调用时才处理实参，它可以接收多个参数或默认参数

规则长、需要多步判断或需要文档字符串时，使用具名 def 函数更容易维护

```python
pairs = [("S01", 7.8), ("S02", 2.7)]
by_mass = sorted(
    pairs, key=lambda pair: pair[1]
)
print(by_mass)
```

```python
values = [0.0, None, 7.8]
show = lambda x: (
    f"{x:.1f} g" if x is not None
    else "缺失"
)
print([show(x) for x in values])
```

```python
def apply_all(values, rule):
    return [rule(x) for x in values]
def kg_to_g(x):
    return x * 1000
result = apply_all([0.2, 0.5], kg_to_g)
print(result)
```

```python
def number_sum(n):
    if n == 0:
        return 0
    return n + number_sum(n - 1)
print(number_sum(3))
```

算法选练：随机撒点估计圆周率

```python
from random import Random
rng = Random(2026)
n, inside = 10_000, 0
for _ in range(n):
    x, y = rng.random(), rng.random()
    inside += x * x + y * y <= 1
print(f"{4 * inside / n:.4f}")
```
